```python
# Student Percentage Checker

try:
    student_marks = float(input("Enter Your percentage: "))

    if student_marks < 0 or student_marks > 100:
        print("Please enter a percentage between 0 and 100.")

    elif student_marks > 90:
        print("Hurray! 🎉 You achieved A1!")

    elif student_marks >= 80:
        print("Excellent! You achieved A2!")

    elif student_marks >= 70:
        print("Very Good! You achieved B1!")

    elif student_marks >= 60:
        print("Good Job! You achieved B2!")

    elif student_marks >= 50:
        print("Good Job! You achieved C1!")

    elif student_marks >= 40:
        print("Good Job! You achieved C2!")

    elif student_marks >= 30:
        print("Good Job! You achieved D1!")

    elif student_marks >= 20:
        print("Padhle chup chap, D2 laya hai, kya krke manega, 'F' lane ka plan hai kya")

    elif student_marks >= 10:
        print("Unpadh hai to!")

    else:
        print("Good Luck for the next time!")

except ValueError:
    print("Invalid input! Please enter a valid number.")
```
